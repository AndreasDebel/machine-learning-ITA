```python
import matplotlib.pyplot as plt
import numpy as np

X = 6 * np.random.rand(100, 1) - 3
y = 0.5 * X * X + X + 2 + np.random.randn(100, 1)

plt.plot(X, y, "g.")
plt.axis([-3, 3, 0, 10])
plt.show()
```

```python
# The x value is used for two thetas (x and x^2)
# ... because the cost function needs to find THREE thetas this time.
from sklearn.preprocessing import PolynomialFeatures

poly_features = PolynomialFeatures(2, include_bias=False)
X_poly = poly_features.fit_transform(X)

print(X[0], X_poly[0])
```

```python
from sklearn.linear_model import LinearRegression

lm = LinearRegression()
lm.fit(X_poly, y)
print(lm.intercept_, lm.coef_)

f = lambda x: lm.coef_[0][1]*x*x + lm.coef_[0][0]*x + lm.intercept_

plt.plot(X, y, "g.")
plt.plot(X, f(X), "b.")
plt.axis([-3, 3, 0, 10])
plt.show()
```

```python
# Now another layey: x^3!

X = 6 * np.random.rand(100, 1) - 3
y = 0.3 * X**3 - 0.5 * X**2 - X + 2 + np.random.randn(100, 1)

poly3 = PolynomialFeatures(3, include_bias=False)
X_poly3 = poly3.fit_transform(X)

lm3 = LinearRegression()
lm3.fit(X_poly3, y)
print(lm3.intercept_, lm3.coef_)

f3 = lambda x: (lm3.coef_[0][2]*x**3 + lm3.coef_[0][1]*x**2
                + lm3.coef_[0][0]*x + lm3.intercept_)

plt.plot(X, y, "g.")
plt.plot(X, f3(X), "b.")
plt.axis([-3, 3, -10, 5])
plt.show()
```
