```python
# Ex 1.2: plot points
import matplotlib.pyplot as plt
import numpy as np

X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

plt.plot(X, y, "b.")
plt.axis([0, 2, 0, 15])
plt.show()
```

```python
print(type(X))
print(X.shape)
print(y.shape)
```

```python
# Ex 1.3 Find a good cost function
def cost(a, b, X, y):
    m = len(y)
    error = a + b*X - y
    return np.sum(error ** 2) / (2*m)

ainterval = np.arange(1, 10, 0.05)
binterval = np.arange(0.5, 5, 0.05)

best = (None, None, float("inf"))
for a in ainterval:
    for b in binterval:
        J = cost(a, b, X, y)
        if J < best[2]:
            best = (a, b, J)
print(best)
```

```python
# Test to see how lines might fit well or bad
plt.plot(X, y, "b.")
xs = np.array([0, 2])
plt.plot(xs, 1 + 1*xs, "r-")   # bad line
plt.plot(xs, 4 + 3*xs, "g-")   # good line
plt.show()

print(cost(1, 1, X, y))   # red/bad line
print(cost(4, 3, X, y))   # green/good line
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Ex 1.4: 3D representation of cost functions




# First plot x and y values via the intervals
ainterval = np.arange(-10, 10, 0.05)   # 400 theta0 values
binterval = np.arange(-10, 10, 0.05)   # 400 theta1 values
A, B = np.meshgrid(ainterval, binterval)

# Then calculate cost for each of the 160.000 points 
zs = np.array([cost(a, b, X, y) for a, b in zip(np.ravel(A), np.ravel(B))])
Z = zs.reshape(A.shape)

# Plot surface in 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(A, B, Z)
ax.set_xlabel('Theta0')
ax.set_ylabel('Theta1')
ax.set_zlabel('Cost')
plt.show()
```
